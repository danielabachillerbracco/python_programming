## 1. List comprehensions: transform

A list comprehension has the form `[expression for item in iterable]`. It creates a new list and leaves the source unchanged.


### Worked example

Predict the result first. Then run the cell and change one input.


```python
prices = [10, 20, 35]
prices_with_tax = [round(price * 1.21, 2) for price in prices]
labels = [f"€{price:.2f}" for price in prices_with_tax]
print(prices, prices_with_tax, labels)
```

### Guided exercise 1

Given monthly revenue, create: revenue in thousands, formatted labels, and month-over-month differences.


```python
revenue = [120000, 135000, 128000, 150000]
# TODO: revenue_k = [...] 
revenue_k = [value / 1000 for value in revenue]
print(revenue_k)
# TODO: labels = [...]
labels = [f'${value/1000:.0f}k' for value in revenue]
# TODO: differences using range or zip
differences = [current-previous for previous, current in zip(revenue, revenue[1:])]
assert revenue_k==[120,135,128,150]
assert labels == ["$120k", "$135k", "$128k", "$150k"]
assert differences == [15000, -7000, 22000]

print(revenue)
```

## 2. List comprehensions: filter and transform

Add an `if` after the iterable to filter. Put a conditional expression before `for` when every input must produce an output.

Predict the result first. Then run the cell and change one input.


```python
amounts = [12, -4, 0, 25, 8]
positive = [x for x in amounts if x > 0]
status = ["positive" if x > 0 else "not positive" for x in amounts]
print(positive, status)
```

### Guided exercise 2

From transactions, create expense amounts, high-value IDs and signed amounts. Keep the three intentions in separate comprehensions.


```python
transactions = [
 {"id":"T1","kind":"income","amount":1000},
 {"id":"T2","kind":"expense","amount":120},
 {"id":"T3","kind":"expense","amount":30},
]
# TODO: expense_amounts
expense_amount = [t["amount"] for t in transactions if t["kind"] == "expense"]

# TODO: high_value_ids (amount >= 100)
high_v= [t["id"] for t in transactions if t["amount"] >= 100]

# TODO: signed_amounts (income positive, expense negative)
signed_amounts = [-t["amount"] if t["kind"] == "expense" else t["amount"] for t in transactions]

print(expense_amount)
print(high_v)
print(signed_amounts)
```

## 3. Nested list comprehensions

A nested comprehension can flatten small, regular nested structures. If the logic needs several conditions or side effects, use an explicit loop.


### Worked example

Predict the result first. Then run the cell and change one input.


```python
teams = [["Ana","Luis"],["Marta"],["Leo","Sara"]]
members = [name for team in teams for name in team]
pairs = [(row, col) for row in range(2) for col in range(3)]
print(members, pairs)
```

## 4. Set comprehensions

A set comprehension uses `{expression for item in iterable}`. It transforms values and removes duplicates. It is ideal for categories, identifiers and validation vocabularies.


### Worked example

Predict the result first. Then run the cell and change one input.


```python
emails = [" Ana@Example.com ","ana@example.com","LUIS@example.com"]
unique_emails = {email.strip().casefold() for email in emails}
domains = {email.split("@")[-1].casefold() for email in unique_emails}
print(unique_emails, domains)
```

### Guided exercise 4

Create unique normalised categories, categories with expense totals above 100, and duplicate IDs detected from the original list.


```python
records = [
 {"id":"T1","category":" Food ","amount":80},
 {"id":"T2","category":"food","amount":40},
 {"id":"T2","category":"Travel","amount":200},
]
# TODO: categories set comprehension
categories = {record["category"].strip().casefold() for record in records}
print(categories)

# TODO: large_categories set comprehension
totals = {}
for record in records:
    cat = record["category"].strip().casefold()
    totals[cat] = totals.get(cat, 0) + record["amount"]
large = {cat for cat in totals if totals[cat] > 100}
print(totals)
print(large)

# TODO: duplicate_ids (a helper count dictionary is acceptable)
id_count = {}
for record in records:
    id_count[record["id"]] = id_count.get(record["id"], 0) + 1
duplicate_ids = {i for i in id_count if id_count[i] > 1}
print(duplicate_ids)
```

## 5. Dictionary comprehensions

A dictionary comprehension uses `{key_expression: value_expression for item in iterable}`. Keys must be unique; later duplicates overwrite earlier values.


### Worked example

Predict the result first. Then run the cell and change one input.


```python
products = [
 {"id":"P1","name":"Keyboard","price":49},
 {"id":"P2","name":"Mouse","price":25},
]
by_id = {product["id"]: product for product in products}
price_by_name = {product["name"]: product["price"] for product in products}
discounted = {key: round(value*0.9,2) for key,value in price_by_name.items()}
print(by_id, price_by_name, discounted)
```

### Guided exercise 5

Build an index by transaction ID, a dictionary of signed amounts, and category totals using a dictionary comprehension plus `sum()`.


```python
transactions = [
 {"id":"T1","kind":"income","category":"salary","amount":1000},
 {"id":"T2","kind":"expense","category":"food","amount":80},
 {"id":"T3","kind":"expense","category":"food","amount":20},
]
# TODO: by_id
by_id = {t["id"]: t for t in transactions}

# TODO: signed_by_id
signed_by_id = {t["id"]: -t["amount"] if t["kind"] == "expense" else t["amount"] for t in transactions}

# TODO: total_by_category
cats = {t["category"] for t in transactions}
total_by_category = {c: sum([t["amount"] for t in transactions if t["category"] == c]) for c in cats}

print(by_id["T2"])
print(signed_by_id)
print(total_by_category)
```

## 6. `enumerate`, `zip` and `sorted`

`enumerate` adds positions, `zip` aligns sequences, and `sorted(key=...)` orders values without changing the source.


### Worked example

Predict the result first. Then run the cell and change one input.


```python
names=["Ana","Luis","Marta"]
scores=[9,7,10]
rows=[{"rank":rank,"name":name,"score":score}
      for rank,(name,score) in enumerate(zip(names,scores),start=1)]
ranking=sorted(rows,key=lambda row:(-row["score"],row["name"]))
print(ranking)
```

### Guided exercise 6

Combine months and revenue, sort by revenue descending, and add a display rank after sorting.


```python
months=["Jan","Feb","Mar"]
revenue=[120,150,135]
# TODO: monthly rows with zip
rows = [{"month": m, "revenue": r} for m, r in zip(months, revenue)]

# TODO: sorted rows
sorted_rows = sorted(rows, key=lambda row: row["revenue"], reverse=True)

# TODO: ranked rows with enumerate
ranked = [{"rank": i, "month": row["month"], "revenue": row["revenue"]} for i, row in enumerate(sorted_rows, start=1)]
print(ranked)
```

## Final guided challenge

Clean a transaction dataset with comprehensions: normalise categories, reject non-positive amounts, index by ID, collect unique categories, calculate category totals and create a stable ranking.

Build it in small steps and keep every previous check passing.


```python
raw = [
 {"id":"T2","category":" Food ","amount":80},
 {"id":"T1","category":"Travel","amount":200},
 {"id":"T3","category":"food","amount":20},
 {"id":"T4","category":"Health","amount":0},
]
# TODO: clean list comprehension
clean = [{"id": r["id"], "category": r["category"].strip().casefold(), "amount": r["amount"]}
         for r in raw if r["amount"] > 0]

# TODO: by_id dict comprehension
by_id = {r["id"]: r for r in clean}

# TODO: categories set comprehension
categories = {r["category"] for r in clean}

# TODO: totals dict comprehension
totals = {c: sum([r["amount"] for r in clean if r["category"] == c]) for c in categories}

# TODO: ranking with sorted and enumerate
ordered = sorted(totals.items(), key=lambda item: (-item[1], item[0]))
ranking = [{"rank": i, "category": c, "total": t} for i, (c, t) in enumerate(ordered, start=1)]

# TODO: checks
assert len(clean) == 3                      
assert "T4" not in by_id
assert by_id["T2"]["category"] == "food"    
assert categories == {"food", "travel"}
assert totals == {"food": 100, "travel": 200}
assert ranking[0] == {"rank": 1, "category": "travel", "total": 200}
assert ranking[1]["category"] == "food"
assert raw[0]["category"] == " Food "  

print(clean)
print(totals)
print(ranking)
```
