## Quick reminders

- End an `if`, `elif`, `else`, `for`, or `while` header with `:`. Indent its body consistently, normally with four spaces.
- `=` assigns a value; `==` compares values. Use `and`, `or`, and `not` to combine or negate conditions.
- `range(start, stop, step)` excludes `stop`.
- `break` exits the nearest loop; `continue` skips to its next iteration.
- A `while` loop needs a way to stop. Update its control variables on every relevant path, especially when using `continue`.
- Check list or string bounds before indexing. In `and` expressions, Python evaluates the right side only if the left side is true.
- If your loop keeps running unexpectedly, interrupt the kernel, fix the condition or update, and rerun the cell to restore its starting values.
