## Quick reminders

- **List:** `[1, 2, 3]` — an ordered, mutable collection that allows duplicates. Indices start at `0`.
- **Set:** `{1, 2, 3}` — a mutable collection of unique elements, without positional indexing. Display order may vary. Use `set()` for an empty set.
- **Dictionary:** `{"name": "Alex"}` — a mutable mapping from unique keys to values. Use `{}` for an empty dictionary.
- Methods such as `.append()`, `.sort()`, and `.update()` modify a collection in place and return `None`. Call the method, then print the collection.
- `.copy()` makes a shallow copy. The copy exercises here use only immutable values.
