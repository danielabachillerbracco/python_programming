## Boundary laboratory
Create at least three tests: normal, boundary, and invalid. State expected behavior before execution.

```python
# TODO 3 — assertions / explicit expected errors
import math
#1. Normal Tests - everyday inputs
#Prediction: each function gives the obvious answer.
assert add(2, 3) == 5 #expected output: 5
assert subtract(10, 4) == 6 #expected output: 6
assert multiply(3, 4) == 12 #expected output: 12
assert divide(6, 3) == 2 #expected output: 2
assert math.isclose(divide(2,3),0.67, abs_tol=0.01) #expected output: 0.67

#2. Boundary Tests - the edges: zero, negative numbers, decimals, large numbers
#Prediction: each function gives the obvious answer.
#2.i. Zero
assert add(0, 0) == 0 #expected output: 0
assert subtract(0, 0) == 0 #expected output: 0
assert multiply(0, 99) == 0 #expected output: 0
assert divide(0, 5) == 0 #expected output: 0
#2.ii. Negative numbers
assert add(-1, 1) == 0 #expected output: 0
assert subtract(2,3) == -1 #expected output: -1
assert multiply(-2, 3) == -6 #expected output: -6
assert multiply(-2, -3) == 6 #expected output: 6
assert divide(-6, 3) == -2 #expected output: -2
#2.iii. Decimals
assert math.isclose(add(0.1, 0.2), 0.3, abs_tol=0.01) #expected output: 0.3
assert math.isclose(multiply(0.5,0.5), 0.25) #expected output: 0.25
#2.iv. Large numbers
assert multiply(1000000,1000000) == 1000000000000 #expected output: 1000000000000

#3. Invalid Tests - inputs that are not numbers, or divide by zero
#Prediction: divide(5,0) raises ValueError
try:
    divide(5,0)
    raise AssertionError("Expected ValueError, but nothing was raised.")
except ValueError as error:
    print("Invalid test passed: divide(5,0) raised ValueError ->", error)
    
#Evidence: the actual output expected
print("add(0.1,0.2) expected 0.3 actual", add(0.1,0.2))
print("divide(2,3) expected 0.67 actual", divide(2,3))
print("multiply(-2,3) expected -6 actual", multiply(-2,3))

print("All tests passed successfully.")
```

## Debugging checkpoint
If a test fails: capture the smallest failing input, exception or actual value, hypothesis, one change, and new evidence.

```python
# TODO 4 — corrected implementation or diagnostic experiment
#Bug A: the test was wrong
# - Failing input: assert divide(2,3)==0.67
# - Actual value: divide (2,3) returns 0.6666666666666666 -> AssertionError
# - Hypothesis: divide is correct but the test is wrong. 2/3 never ends in decimal and == compares every digit, so 0.6666666666666666 != 0.67
# - One change: compare with math.isclose() instead of ==, which allows for a tolerance in the comparison. (no change in calculator.py)
# - New evidence: see below:
actual = divide(2,3)
print("Actual value of divide(2,3) is", actual)

#Evidence before the change: the old test fails:
try:
    assert actual == 0.67
    print("Old test passed unexpectedly.")
except AssertionError:
    print("Old test with == fails, as observed")
    
#Evidence after the change: the new test passes:
assert math.isclose(actual, 0.67, abs_tol=0.01)
print("New test with math.isclose() passes, as expected.")


#Bug B: the function was wrong
# - Smallest failing input: divide(5,0)
# - Exception: ZeroDivisionError (Python's built-in exception) but the specification required a ValueError.
# - Hypothesis: the first version of the divide function had no guard clause, it went directly to a/b.
# - One change: added to src/calculator.py:
# if b==0:
#    raise ValueError("Divisor cannot be zero.")
# - New evidence: see below:
try:
    divide(5,0)
    raise AssertionError("Expected ValueError, but nothing was raised.")
except ValueError as error:
    print("Corrected divide(5,0) raises ValueError ->", error)
```

## Stretch challenge
Add percentage change and compound growth operations with explicit edge-case policies.

```python
# OPTIONAL — stretch solution
import math
from calculator import percentage_change, compound_growth
#Policies:
#percentage_change: old == 0->ValueError (change from zero is undefined)
#compound_growth: negative rate allowed (a decline), down to -1 (total loss)
#                 rate below -1 -> ValueError
#                 periods==0->principal unchanged
#                 periods <0 -> ValueError
#Normal and Boundary Tests for percentage_change
assert math.isclose(percentage_change(100,110),10) #expected output: 10 (10%)
assert math.isclose(percentage_change(50,100),100) #expected output: 100 (100%) doubled
assert percentage_change(200,150) == -25 #expected output: -25 (25% decrease)
assert percentage_change(80,80) == 0 #expected output: 0 (no change)
#Normal and Boundary Tests for compound_growth
assert math.isclose(compound_growth(1000,0.05,2), 1102.5) #expected output: 1102.5 (normal case)
assert math.isclose(compound_growth(1000, -0.10, 1),900) #expected output: 900 (negative rate allowed)
assert compound_growth(1000, -1, 3) == 0 #expected output: 0 (total loss)
assert compound_growth(1000, 0.05, 0) == 1000 #expected output: 1000 (periods==0)
assert compound_growth(1000, 0, 5) == 1000 #expected output: 1000 (rate==0)
#Invalid tests (the errors)
try:
    percentage_change(0,50)
    raise AssertionError("Expected ValueError, but nothing was raised.")
except ValueError as error:
    print("Invalid test passed: percentage_change(0,50) ->", error)
try:
    compound_growth(1000, -1.5, 2)
    raise AssertionError("Expected ValueError, but nothing was raised.")
except ValueError as error:
    print("Invalid test passed: compound_growth(1000, -1.5, 2) ->", error)
try:
    compound_growth(1000, 0.05, -1)
    raise AssertionError("Expected ValueError, but nothing was raised.")
except ValueError as error:
    print("Invalid test passed: compound_growth(1000, 0.05, -1) ->", error)
#Evidence of the actual output expected
print("compound_growth(1000,0.05,2) expected 1102.5 actual", compound_growth(1000,0.05,2))
print("percentage_change(100,110) expected 10 actual", percentage_change(100,110))
print ("All stretch tests passed successfully.")
```

I refused rates below 100% because the formula returns a believable but meaningless number, a wrong number that looks real is more dangerous in a report than an error
