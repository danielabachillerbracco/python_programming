## Syntax reminders

- **List:** `[expression for item in iterable]`
- **Set:** `{expression for item in iterable}`
- **Dictionary:** `{key_expression: value_expression for item in iterable}`
- **Filter:** place `if condition` after the `for` clause to keep only matching items.
- **Choose a value:** place `value_if_true if condition else value_if_false` before the `for` clause to produce one result for every item.
- With two `for` clauses, the first is the outer loop and the second is the inner loop.
- Lists preserve iteration order and allow duplicate results. Sets keep unique values; their display order may vary.
- Dictionary keys are unique. When a key is produced more than once, the last value assigned to it is retained.
- Set elements and dictionary keys must be hashable. Numbers, strings, and tuples containing only hashable values work; lists do not.
